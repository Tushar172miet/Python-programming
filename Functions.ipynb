```python
#Program to generate fisrt n terms of fibonacci series using functions
def print_fibonacci(n):
    # Start with the first two numbers
    a, b = 0, 1

    # Loop n times to print each number
    for _ in range(n):
        print(a, end=" ")
        # Update a and b to the next numbers
        a, b = b, a + b

# --- Run the function ---

print_fibonacci(7)
```

```python
#write a program to generate end terms in fibonacci series using recursion
def fibonacci(n):
  if n <= 0:
    return []
  elif n == 1:
    return [0]
  elif n == 2:
    return [0, 1]
  else:
    fib_series = fibonacci(n - 1)
    fib_series.append(fib_series[-1] + fib_series[-2])
    return fib_series
print(fibonacci(7))
```

```python
#to find gcd of a numbers using recursion
def gcd(a,b):
  if b == 0:
    return a
  else:
    return gcd(b, a % b)

print(gcd(12,14))
```

```python
#to find armstrong number using recursion
def is_armstrong(n):
  num_str = str(n)
  num_digits = len(num_str)
  total_sum = sum(int(digit) ** num_digits for digit in num_str)
  return total_sum == n
print(is_armstrong(153))
```

```python
def reverse(num):
  rev = 0
  while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num //= 10
  return rev

print(reverse(123))
```
